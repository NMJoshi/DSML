```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("./data/train.csv")
```

```python
df.head(5)
```

```python
df.info()
```

```python
df=df.drop(columns=['id'])
```

```python
# checking null values
df.isnull().sum()
```

```python
df.describe(include="all").T
```

```python
vehicle_age_dict = {"< 1 Year":0,"1-2 Year":0,"> 2 Years":1}
df["Vehicle_Age"]=df["Vehicle_Age"].map(vehicle_age_dict)
```

```python
df.head(5)
```

```python
plt.bar(df["Age"],df["Response"])
plt.xlabel("Vehicle_Age")
plt.ylabel("Response")
plt.title(f"Vehical Age to Response")
plt.xticks(ticks=[0,1],rotation=0)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.figure(figsize=(10,6))
plt.show()
```
